## load_and_save

```python
import pandas as pd
```

## read_csv

```python
df = pd.read_csv(filepath_or_buffer="../data/feed-views.log",
                 sep='\t',
                 header=None,
                 skiprows=[2,3],
                 skipfooter=2,
                 names=['datetime', 'user'],
                 index_col='datetime',
                 engine='python')
```

## Rename datetime to date_time

```python
df.index.name = 'date_time'
```

## to_csv

```python
df.to_csv(sep=";", path_or_buf = "../data/feed-views-semicolon.log")
```
