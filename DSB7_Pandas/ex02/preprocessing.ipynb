## preprocessing

```python
import pandas as pd
```

## Make ID the index column

```python
df = pd.read_csv('../data/auto.csv',
                 sep=',',
                 index_col='ID')
```

## Count the number of observations

```python
df.count()
```

## Drop the duplicates

```python
df.drop_duplicates(['CarNumber', 'Make_n_model', 'Fines'], inplace=True, keep='last')
df.count()
```

## Work with missing values

```python
df.isnull().sum()
```

```python
df.dropna(axis=1, thresh=500, inplace=True)
df.isnull().sum()
```

```python
df['Refund'] = df['Refund'].ffill()
df.isnull().sum()
```

```python
mean = df['Fines'].mean()
df['Fines'] = df['Fines'].fillna(mean)
df.isnull().sum()
```

## Split and parse the make and model

```python
def split(text):
    row = text.split(' ')
    return pd.Series(row)
df[['Make', 'Model']] = df['Make_n_model'].apply(split)
df.drop(labels='Make_n_model', axis=1, inplace=True)
df.to_json('../data/auto.json', orient='records')
```
