## selects_n_aggs

```python
import pandas as pd
```

## Load the JSON file

```python
df = pd.read_json("../data/auto.json")
df.set_index('CarNumber', inplace=True)
```

## Display the rows where the fines are greater than 2,100

```python
df[df['Fines'] > 2100]
```

## Display the rows where the fines are greater than 2,100 and the refund equals 2

```python
df[(df["Fines"] > 2100) & (df["Refund"] == 2)]
```

## Display the rows where the models are from the list: [’Focus’, ’Corolla’]

```python
models = ['Focus', 'Corolla']
df.loc[df['Model'].isin(models)]
```

## Display the rows where the car number is from the list:

```python
cars_numbers = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
df[df.index.isin(cars_numbers)]
```

## Display the median fines grouped by make

```python
df.groupby('Make')['Fines'].median()
```

## Display the median fines grouped by make and model

```python
df.groupby(['Make', 'Model'])['Fines'].median()
```

## Display the number of fines grouped by make and model

```python
df.groupby(['Make', 'Model'])['Fines'].count()
```

## Display the minimum and maximum fines grouped by make and model

```python
df.groupby(['Make', 'Model'])['Fines'].min()
```

```python
df.groupby(['Make', 'Model'])['Fines'].max()
```

## Display the standard deviation of fines grouped by make and model

```python
df.groupby(['Make', 'Model'])['Fines'].std()
```

## Display the car numbers, grouped by the number of fines in descending order

```python
cars_numbers_count = pd.DataFrame(df.groupby(df.index).count()['Fines'])
cars_numbers_count.sort_values('Fines', ascending=False, inplace=True)
cars_numbers_count
```

## Select from the initial dataframe all the rows corresponding to the top-1 car number

```python
df[df.index == cars_numbers_count.index[0]]
```

## Display the car numbers grouped by sum of fines in descending order

```python
cars_numbers_sum = pd.DataFrame(df.groupby(df.index).sum()['Fines'])
cars_numbers_sum.sort_values('Fines', ascending=False, inplace=True)
cars_numbers_sum
```

## Select all the rows corresponding to the top-1 car number from the initial dataframe

```python
df[df.index == cars_numbers_sum.index[0]]
```

## Display a table that answers the question

```python
pd.DataFrame(df.groupby(df.index)['Model'].nunique()).sort_values('Model', ascending=False)
```
