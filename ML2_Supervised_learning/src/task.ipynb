## Derive an analytical solution to the regression problem. Use a vector form of the equation.

**1. Обозначения**

Пусть у нас есть $N$ наблюдений и $D$ признаков.

- y ∈ $\mathbb{R}^N$ — вектор целевых значений (размер $N×1$)
- $\theta∈\mathbb{R}^{D+1}$ — вектор параметров модели, включающий $b$ и все веса $w_1,\dots,w_D$
- $X$ — матрица признаков размера $N×(D+1)$. Первый столбец состоит из единиц и нужен для учёта свободного члена $b$

Вектор параметров $\theta$:

$$
\theta = \begin{pmatrix}
b \\
w_1 \\
w_2 \\
\dots \\
w_D
\end{pmatrix}
$$

Матрица признаков $X$:

$$
X = \begin{pmatrix}
1 & x_{11} & x_{12} & \dots & x_{1D} \\
1 & x_{21} & x_{22} & \dots & x_{2D} \\
\dots & \dots & \dots & \dots & \dots \\
1 & x_{N1} & x_{N2} & \dots & x_{ND}
\end{pmatrix}
$$

Модель:

$$
\hat{y} = X\theta = \begin{pmatrix}
b + w_1 x_{11} + w_2 x_{12} + \dots + w_D x_{1D} \\
b + w_1 x_{21} + w_2 x_{22} + \dots + w_D x_{2D} \\
\dots \\
b + w_1 x_{N1} + w_2 x_{N2} + \dots + w_D x_{ND}
\end{pmatrix}
$$

**2. Функция потерь**

Среднеквадратичная ошибка в векторной форме:

$$MSE(\theta) = \frac{1}{N}(y-X\theta)^T(y-X\theta)$$

Множитель $\frac{1}{N}$ не влияет на положение минимума, поэтому будем минимизировать сумму квадратов ошибок:

$$S(\theta) = (y-X\theta)^T(y-X\theta)$$

**3. Пошаговое дифференцирование**

Перед расчётом зафиксируем ключевые правила (для векторов и матриц):

- $\nabla_\theta (a^T \theta) = a$ — производная линейной формы;
- $\nabla_\theta (\theta^T A \theta) = (A + A^T)\theta$ — производная квадратичной формы;
- Если $A$ симметрична ($A = A^T$), то $\nabla_\theta (\theta^T A \theta) = 2A\theta$;
- $\nabla_\theta c = 0$, если $c$ не зависит от $\theta$.

Исходная функция:

$$S(\theta) = (y - X\theta)^T (y - X\theta)$$

**Шаг 1. Раскрытие скобок**

Раскроем выражение:

$$S(\theta) = y^T y - y^T X\theta - \theta^T X^T y + \theta^T X^T X\theta$$

Заметим, что $y^T X\theta$ и $\theta^T X^T y$ — скаляры и равны друг другу (поскольку транспонирование скаляра не меняет его значения):

$$y^T X\theta = \theta^T X^T y$$

Упрощаем:

$$S(\theta) = y^T y - 2\theta^T X^T y + \theta^T X^T X\theta$$

**Шаг 2. Дифференцирование каждого слагаемого**

**Первое слагаемое:** $y^T y$

- не зависит от $\theta$;
- по правилу 4: $\nabla_\theta (y^T y) = 0$.

**Второе слагаемое:** $-2\theta^T X^T y$

- это линейная форма относительно $\theta$;
- представим как $-2(\theta^T a)$, где $a = X^T y$;
- по правилу 1: $\nabla_\theta (\theta^T a) = a$;
- с учётом множителя $-2$: $\nabla_\theta (-2\theta^T X^T y) = -2X^T y$.

**Третье слагаемое:** $\theta^T X^T X\theta$

- это квадратичная форма $\theta^T A\theta$, где $A = X^T X$;
- матрица $A = X^T X$ всегда симметрична (поскольку $(X^T X)^T = X^T X$);
- применяем правило 3: $\nabla_\theta (\theta^T A\theta) = 2A\theta$;
- подставляем $A = X^T X$: $\nabla_\theta (\theta^T X^T X\theta) = 2X^T X\theta$.

**Шаг 3. Суммирование результатов**

Складываем градиенты всех слагаемых:

$$\nabla_\theta S(\theta) = \nabla_\theta (y^T y) + \nabla_\theta (-2\theta^T X^T y) + \nabla_\theta (\theta^T X^T X\theta)$$

Подставляем найденные значения:

$$\nabla_\theta S(\theta) = 0 - 2X^T y + 2X^T X\theta$$

Выносим общий множитель $-2X^T$:

$$\nabla_\theta S(\theta) = -2X^T (y - X\theta)$$

**Шаг 4. Поиск минимума**

Раскроем скобки:

$$X^Ty-X^TX\theta = 0$$

Получаем систему нормальных уравнений:

$$X^TX\theta = X^Ty$$

**4. Аналитическое решение**

Если матрица $X^TX$ обратима, решение имеет вид:

$$\boxed{\theta=(X^TX)^{-1}X^Ty}$$

Это и есть аналитическое решение задачи линейной регрессии методом наименьших квадратов.

**5. Примечание**

- Решение существует и единственно, если столбцы матрицы $X$ линейно независимы (нет строгой мультиколлинеарности)
- При нарушении условия обратимости применяют регуляризацию (Lasso, Ridge) или псевдообратную матрицу

## What changes in the solution when L1 and L2 regularizations are added to the loss function.


В общем виде регуляризованная функция потерь содержит дополнительный штрафной член $R(\theta)$:
$$\min_\theta \sum_{i=1}^N L\bigl(f(x_i,\theta), y_i\bigr) + \lambda R(\theta)$$

**L2-регуляризация (Ridge):** $R(\theta) = \|\theta\|_2^2 = \sum_{i=1}^d \theta_i^2$

Аналитическое решение принимает вид:
$$\theta = (X^T X + \lambda I)^{-1} X^T y$$
Матрица $X^T X + \lambda I$ всегда обратима (даже при мультиколлинеарности). Веса сжимаются к нулю, но **не обнуляются**.

**L1-регуляризация (Lasso):** $R(\theta) = \|\theta\|_1 = \sum_{i=1}^d |\theta_i|$

Аналитического решения нет. Из-за модуля в штрафе многие коэффициенты становятся **точно равными нулю** (разреженность). Оптимум ищут итеративно (субградиентные методы, coordinate descent).

## Explain why L1 regularization is often used to select features. Why are there many weights equal to 0 after the model is fit?


L1-регуляризация добавляет штраф $R(\theta) = \|\theta\|_1 = \sum_{i=1}^d |\theta_i|$.
Каждый ненулевой коэффициент увеличивает штраф. Если вклад некоторого признака в уменьшение ошибки мал (меньше $\lambda$), оптимальным становится $\theta_j = 0$.
Таким образом, L1 автоматически приводит к разреженным решениям и может использоваться для **отбора признаков**: неинформативные признаки получают нулевые веса.

## Explain how you can use the same models (Linear regression, Ridge, etc.) but make it possible to fit nonlinear dependencies.

Модель остаётся **линейной по параметрам**, но признаки предварительно преобразуются нелинейно.
Вместо $X$ используется матрица $\Phi = \phi(X)$, где $\phi$ – нелинейное отображение.

**Примеры преобразований**

$$\phi(x) = (x, x^2, x^3, \dots, x^d)$$

или

$$\phi(x_1,x_2) = (x_1, x_2, x_1x_2, x_1^2, x_2^2, \dots)$$

**Модель с преобразованными признаками**

$$\hat{y} = \Phi\theta$$

Все методы (аналитическое решение OLS, SGD, Ridge, Lasso) применяются к $\Phi$ без изменений. Регуляризация особенно важна, так как большое число полиномиальных признаков легко ведёт к переобучению.

# 4. Models implementation — Linear regression

## Implement a Python class for a linear regression algorithm with two basic methods — fit and predict. Use stochastic gradient descent (SGD) to find optimal model weights. For better understanding, we recommend implementing separate versions of the algorithm with the analytical solution and non-stochastic gradient descent under the hood.

```python
class MyLinearRegression:
    def __init__(self, learning_rate=0.01, epochs=100, random_state=42):
        self.lr = learning_rate
        self.epochs = epochs
        self.random_state = random_state
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        np.random.seed(self.random_state)
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        X_vals = X.values if hasattr(X, 'values') else X
        y_vals = y.values if hasattr(y, 'values') else y

        for _ in range(self.epochs):
            for i in range(n_samples):
                idx = np.random.randint(0, n_samples)
                x_i = X_vals[idx]
                y_i = y_vals[idx]

                y_pred = np.dot(x_i, self.weights) + self.bias
                error = y_pred - y_i

                self.weights -= self.lr * (2 * x_i * error)
                self.bias -= self.lr * (2 * error)

    def predict(self, X):
        X_vals = X.values if hasattr(X, 'values') else X
        return np.dot(X_vals, self.weights) + self.bias
```

## What is determenistic model? Make SGD determenistic.

Детерминированная модель — модель, которая при одних и тех же входных данных и начальных условиях всегда выдаёт одинаковый результат.
Для детерминированности SGD мы фиксируем `random_state` и устанавливаем `np.random.seed(self.random_state)` в начале `fit`.

## Define the R squared (R2) coefficient and implement a function to calculate it.

```python
def r2_score_custom(y_true, y_pred):
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return 1 - (ss_res / ss_tot)
```

# 5. Regularized models implementation — Ridge, Lasso, ElasticNet

## Implement Ridge, Lasso, ElasticNet algorithms: extend the loss function with L2, L1 and both regularizations accordingly.

```python
class MyRegularizedRegression:
    def __init__(self, lr=0.001, epochs=100, alpha=0.1, l1_ratio=0.5, reg_type='ridge', random_state=42):
        self.lr = lr
        self.epochs = epochs
        self.alpha = alpha        # сила регуляризации
        self.l1_ratio = l1_ratio  # только для elastic
        self.reg_type = reg_type  # 'ridge', 'lasso', 'elastic'
        self.random_state = random_state
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        np.random.seed(self.random_state)
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        X_vals = X.values if hasattr(X, 'values') else X
        y_vals = y.values if hasattr(y, 'values') else y

        for _ in range(self.epochs):
            for i in range(n_samples):
                idx = np.random.randint(0, n_samples)
                x_i = X_vals[idx]
                y_i = y_vals[idx]

                y_pred = np.dot(x_i, self.weights) + self.bias
                error = y_pred - y_i

                # Градиент ошибки без регуляризации
                grad_w = 2 * x_i * error
                grad_b = 2 * error

                # Добавляем штраф
                if self.reg_type == 'ridge':
                    grad_w += 2 * self.alpha * self.weights
                elif self.reg_type == 'lasso':
                    grad_w += self.alpha * np.sign(self.weights)
                elif self.reg_type == 'elastic':
                    grad_w += (self.alpha * self.l1_ratio * np.sign(self.weights) +
                               2 * self.alpha * (1 - self.l1_ratio) * self.weights)

                self.weights -= self.lr * grad_w
                self.bias -= self.lr * grad_b

    def predict(self, X):
        X_vals = X.values if hasattr(X, 'values') else X
        return np.dot(X_vals, self.weights) + self.bias
```

## Make predictions with your algorithm and estimate the model with MAE, RMSE and R2 metrics.

```python
my_ridge = MyRegularizedRegression(reg_type='ridge', alpha=0.1)
my_lasso = MyRegularizedRegression(reg_type='lasso', alpha=0.1)
my_elastic = MyRegularizedRegression(reg_type='elastic', alpha=0.1, l1_ratio=0.5)

my_ridge.fit(X_train_scaled, y_train)
my_lasso.fit(X_train_scaled, y_train)
my_elastic.fit(X_train_scaled, y_train)

for name, model in [('my_ridge', my_ridge), ('my_lasso', my_lasso), ('my_elastic', my_elastic)]:
    tr_pred = model.predict(X_train_scaled)
    te_pred = model.predict(X_test_scaled)
    print(f"{name}:")
    print(f"  Train MAE: {mean_absolute_error(y_train, tr_pred):.2f}")
    print(f"  Test MAE:  {mean_absolute_error(y_test, te_pred):.2f}")
    print(f"  Train R2:  {r2_score_custom(y_train, tr_pred):.4f}")
    print(f"  Test R2:   {r2_score_custom(y_test, te_pred):.4f}\n")
    add_to_tables(name, y_train, tr_pred, y_test, te_pred)
```

# 6. Feature normalization

## First, write several examples of why and where feature normalization is mandatory and vice versa.

**Когда нормализация обязательна:**
- Линейные модели с градиентным спуском (Linear Regression, Logistic Regression, Neural Networks) – без масштабирования сходимость медленная или отсутствует.
- Алгоритмы, основанные на расстоянии (KNN, SVM, K-Means) – доминирование признака с большим масштабом.
- Регуляризованные модели (Ridge, Lasso) – штраф зависит от масштаба коэффициентов.

**Когда нормализация не обязательна:**
- Деревья решений и ансамбли (Random Forest, XGBoost, LightGBM) – они работают с порогами, масштаб не важен.

## Let's consider the first of the classical normalization methods — MinMaxScaler. Write a mathematical formula for this method.

Формула MinMaxScaler:
$$
X_{\text{norm}} = \frac{X - X_{\min}}{X_{\max} - X_{\min}}
$$

## Implement your own function or class for MinMaxScaler feature normalization.

```python
class MyMinMaxScaler:
    def __init__(self):
        self.min = None
        self.max = None

    def fit(self, X):
        self.min = np.min(X, axis=0)
        self.max = np.max(X, axis=0)

    def transform(self, X):
        return (X - self.min) / (self.max - self.min)

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)
```

## Initialize MinMaxScaler() from sklearn.preprocessing.

```python
from sklearn.preprocessing import MinMaxScaler

data_sample = X_train[['bathrooms', 'bedrooms']].values[:5]

my_scaler = MyMinMaxScaler()
my_scaled = my_scaler.fit_transform(data_sample)

sk_scaler = MinMaxScaler()
sk_scaled = sk_scaler.fit_transform(data_sample)

print("Разница между MyMinMaxScaler и sklearn MinMaxScaler:", np.sum(np.abs(my_scaled - sk_scaled)))
```

## Repeat the steps from b to e for another normalization method StandardScaler.

```python
class MyStandardScaler:
    def __init__(self):
        self.mean = None
        self.std = None

    def fit(self, X):
        self.mean = np.mean(X, axis=0)
        self.std = np.std(X, axis=0)

    def transform(self, X):
        return (X - self.mean) / self.std

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)

my_std_scaler = MyStandardScaler()
my_std_scaled = my_std_scaler.fit_transform(data_sample)

sk_std_scaler = StandardScaler()
sk_std_scaled = sk_std_scaler.fit_transform(data_sample)

print("Разница между MyStandardScaler и sklearn StandardScaler:", np.sum(np.abs(my_std_scaled - sk_std_scaled)))
```
