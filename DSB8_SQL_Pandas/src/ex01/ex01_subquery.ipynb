## ex01_subquery.ipynb

```python
import pandas as pd
import sqlite3
```

## Create a connection to the database using the sqlite3 library.

```python
db_path = "../data/checking-logs.sqlite"
connection = sqlite3.connect(db_path)
```

## Get the schema of the checker table.

```python
query = "PRAGMA table_info(checker)"
schema = pd.io.sql.read_sql(query, connection)
```

## Get only the first ten rows of the table to see what the table looks like.

```python
query = "SELECT * FROM checker LIMIT 10;"
first_ten_rows = pd.io.sql.read_sql(query, connection)
display(first_ten_rows)
```

## Count how many rows satisfy the following conditions using one query with any number of subqueries.

```python
query = """
        SELECT COUNT(*) AS cnt
        FROM pageviews
        WHERE uid IN (
            SELECT uid
            FROM checker
            WHERE status = 'ready'
              AND numTrials = 1
              AND labname IN ('laba04', 'laba04s', 'laba05', 'laba06', 'laba06s', 'project1')
        );
        """

cnt = pd.io.sql.read_sql(query, connection)
display(cnt)
```

## Close the connection.

```python
connection.close()
```
