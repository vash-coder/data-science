## ex00_first_select.ipynb

```python
import pandas as pd
import sqlite3
```

## Create a connection to the database using the sqlite3 library.

```python
db_path = '../data/checking-logs.sqlite'
connection = sqlite3.connect(db_path)
```

## Get the schema of the checker table.

```python
query = "PRAGMA table_info(pageviews);"
schema = pd.io.sql.read_sql(query, connection)
display(schema)
```

## Get only the first ten rows of the table to see what the table looks like.

```python
query = "SELECT * FROM pageviews LIMIT 10"
first_ten_rows = pd.io.sql.read_sql(query, connection)
display(first_ten_rows)
```

## Use one query to get the subtable.

```python
query = """
        SELECT
            uid,
            datetime
        FROM pageviews
        WHERE uid LIKE 'user_%'
        ORDER BY uid ASC;
        """
pageviews = pd.io.sql.read_sql(query, connection, index_col='datetime', parse_dates=['datetime'], dtype={'uid': 'object'})
display(pageviews)
```

## Close the connection to the database.

```python
connection.close()
```
