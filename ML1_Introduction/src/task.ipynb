# 1. Intro

## To get started, please write 5 examples of the application of ML methods in life. What is the benefit of using machine learning methods in each of your examples?

1. На электронную почту пришло письмо об результат розыгрыша. Необходимо определить к какой категории необходимо отнести данное письмо. («спам» / «не спам») [Классификация]

_**Преимущество ML**
Невозможно заранее прописать все правила, по которым письмо считается спамом (они постоянно меняются, тексты разные, есть обход фильтров). Модель машинного обучения обучается на реальных примерах и способна адаптироваться к новым типам спама без ручного переписывания правил._

2. Предсказание стоимости акции компании. [Регрессия]

_**Преимущество ML**
Зависимость цены от факторов (новости, рынок, поведение инвесторов) сложная и нелинейная. Обычный алгоритм не сможет описать все эти зависимости, а ML способен выявлять скрытые закономерности в данных и учитывать множество факторов одновременно._

3. Сгруппировать отзывы на сайте по темам. [Кластеризация]

_**Преимущество ML**
Нет заранее известных классов и правил группировки. ML автоматически находит структуру в данных и выделяет группы без разметки, что невозможно сделать вручную на больших объемах текста._

4. Проанализировать юридический документ и сделать выжимку. [Снижение размерности]

_**Преимущество ML**
Документы содержат большое количество информации, и сложно вручную определить, какие части наиболее важны. ML может выявлять ключевые смысловые компоненты и сокращать текст, сохраняя основную информацию._

5. Обучить беспилотный автомобиль ездить по дорогам. [Обучение с подкреплением]

_**Преимущество ML**
Невозможно заранее прописать правила для всех дорожных ситуаций. Обучение с подкреплением позволяет системе учиться и адаптироваться к новым условиям среды._

При многоклассовой классификации каждый объект относится ровно к одному из нескольких взаимоисключающих классов.

При многометочной классификации объект может одновременно принадлежать нескольким классам, и эти классы не являются взаимоисключающими.

## To train the model here, we will not use your new features. Remember this example and use it in Lecture 2. To train the model, we will only consider the features 'bathrooms' and 'bedrooms'.

## Read this Sklearn [info about PolynomialFeatures.](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.PolynomialFeatures.html)

## To use PolynomialFeatures, we first need to split the data into training and test samples. We have already done this for you, please read the training and test data.

```python
train_df = pd.read_json('../datasets/train.json')
test_df = pd.read_json('../datasets/test.json')

p1 = train_df['price'].quantile(0.01)
p99 = train_df['price'].quantile(0.99)

train_df_clean = train_df[(train_df['price'] >= p1) & (train_df['price'] <= p99)].copy()
test_df_clean = test_df[(test_df['price'] >= p1) & (test_df['price'] <= p99)].copy()

features = ['bathrooms', 'bedrooms']
target = 'price'

X_train = train_df_clean[features]
y_train = train_df_clean[target]

X_test = test_df_clean[features]
y_test = test_df_clean[target]
```

## Initialize PolynomialFeatures() with a degree of 10.

```python
poly = PolynomialFeatures(degree=2, include_bias=False)
```

## Apply PolynomialFeatures() to fit and transform your training and test data.

```python
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

print(f"Исходных признаков: {X_train.shape[1]}")
print(f"После PolynomialFeatures (degree=10): {X_train_poly.shape[1]} признаков")
```

# Now you need to train 3 models: linear regression, decision tree and naive model. We will use them as black boxes without deep understanding.

## Results table

### Create two empty Pandas DataFrames with columns 'model', 'train', 'test'. Let's call the first one result_MAE and the second one result_RMSE. We will fill these tables with the results of the models.

```python
result_MAE = pd.DataFrame(columns=['model', 'train', 'test'])
result_RMSE = pd.DataFrame(columns=['model', 'train', 'test'])
```

## Linear Regression

### Initialize linear regression from sklearn with no parameters.

```python
lr = LinearRegression()
lr.fit(X_train_poly, y_train)
```

### Fit your model and make predictions on training and test features. Save it as new columns in data.

```python
y_train_pred_lr = lr.predict(X_train_poly)
y_test_pred_lr = lr.predict(X_test_poly)
```

### Compute MAE (Mean Absolute Error) on training and test targets.

```python
mae_train_lr = mean_absolute_error(y_train, y_train_pred_lr)
mae_test_lr = mean_absolute_error(y_test, y_test_pred_lr)
```

### Calculate RMSE (Root Mean Square Error) on training and test objectives.

```python
rmse_train_lr = np.sqrt(mean_squared_error(y_train, y_train_pred_lr))
rmse_test_lr = np.sqrt(mean_squared_error(y_test, y_test_pred_lr))
```

### Insert your metrics into tables result_MAE and result_RMSE with model name 'linear_regression'.

```python
result_MAE = pd.concat([result_MAE, pd.DataFrame([{
    'model': 'linear_regression',
    'train': mae_train_lr,
    'test': mae_test_lr
}])], ignore_index=True)
```

```python
result_RMSE = pd.concat([result_RMSE, pd.DataFrame([{
    'model': 'linear_regression',
    'train': rmse_train_lr,
    'test': rmse_test_lr
}])], ignore_index=True)
```

## Decision Tree

### Initialize decision tree regressor from sklearn with fixed random_state=21.

```python
dt = DecisionTreeRegressor(random_state=21)
dt.fit(X_train_poly, y_train)
```

### Fit it to train features and train target and make prediction on train and test features. Save it as a new column in data.

```python
y_train_pred_dt = dt.predict(X_train_poly)
y_test_pred_dt = dt.predict(X_test_poly)
```

### Compute MAE (Mean Absolute Error) on train and test targets.

```python
mae_train_dt = mean_absolute_error(y_train, y_train_pred_dt)
mae_test_dt = mean_absolute_error(y_test, y_test_pred_dt)
```

### Compute RMSE (Root Mean Square Error) on train and test targets.

```python
rmse_train_dt = np.sqrt(mean_squared_error(y_train, y_train_pred_dt))
rmse_test_dt = np.sqrt(mean_squared_error(y_test, y_test_pred_dt))
```

### Insert your metrics into tables result_MAE and result_RMSE with model name 'decision_tree'.

```python
result_MAE = pd.concat([result_MAE, pd.DataFrame([{
    'model': 'decision_tree',
    'train': mae_train_dt,
    'test': mae_test_dt
}])], ignore_index=True)
```

```python
result_RMSE = pd.concat([result_RMSE, pd.DataFrame([{
    'model': 'decision_tree',
    'train': rmse_train_dt,
    'test': rmse_test_dt
}])], ignore_index=True)
```

## Naive Models

### Calculate the mean and median of 'price' on the training and test data and create a column with these values.

```python
mean_price = y_train.mean()
median_price = y_train.median()

print(f'Средняя цена (train): {mean_price:.2f}')
print(f'Медианная цена (train): {median_price:.2f}')

y_train_pred_mean = np.full_like(y_train, mean_price)
y_test_pred_mean  = np.full_like(y_test, mean_price)

y_train_pred_median = np.full_like(y_train, median_price)
y_test_pred_median  = np.full_like(y_test, median_price)
```

### Insert your metrics into tables result_MAE and result_RMSE with model names 'naive_mean' and 'naive_median'.

```python
result_MAE = pd.concat([result_MAE, pd.DataFrame([
        {'model': 'naive_mean',   'train': mae_train_mean,   'test': mae_test_mean},
        {'model': 'naive_median', 'train': mae_train_median, 'test': mae_test_median}
])], ignore_index=True)
```
